```python
import pandas as pd

url = 'https://github.com/DataTalksClub/nyc-tlc-data/releases/download/yellow/yellow_tripdata_2021-01.csv.gz'

null_counts = {}
min_vals = {}
max_vals = {}
dtypes = {}
total_rows = 0

print (...)


chunk_iter = pd.read_csv(url, chunksize=100000, low_memory=False)

for chunk in chunk_iter:
    total_rows += len(chunk)
    
    for col in chunk.columns:
        if col not in dtypes:
            dtypes[col] = chunk[col].dtype
        
        null_counts[col] = null_counts.get(col, 0) + chunk[col].isna().sum()
        
        valid_data = chunk[col].dropna()
        if not valid_data.empty:
            try:
                c_min = valid_data.min()
                c_max = valid_data.max()
                min_vals[col] = min(min_vals.get(col, c_min), c_min)
                max_vals[col] = max(max_vals.get(col, c_max), c_max)
            except:
                pass

profile_df = pd.DataFrame({
    'dtype': dtypes,
    'nulls_count': null_counts,
    'nulls_%': {col: round((null_counts[col] / total_rows) * 100, 2) for col in dtypes},
    'min': min_vals,
    'max': max_vals
})

print(f"\nTotal Rows: {total_rows}")
print(profile_df)
```

```python
import pandas as pd

url = 'https://github.com/DataTalksClub/nyc-tlc-data/releases/download/yellow/yellow_tripdata_2021-01.csv.gz'

dtype = {
    "VendorID": "Int64", "passenger_count": "Int64", "trip_distance": "float64", 
    "RatecodeID": "Int64", "store_and_fwd_flag": "string", "PULocationID": "Int64", 
    "DOLocationID": "Int64", "payment_type": "Int64", "fare_amount": "float64", 
    "extra": "float64", "mta_tax": "float64", "tip_amount": "float64", 
    "tolls_amount": "float64", "improvement_surcharge": "float64", 
    "total_amount": "float64", "congestion_surcharge": "float64"
}
parse_dates = ["tpep_pickup_datetime", "tpep_dropoff_datetime"]

print("Loading full dataset for problem counting...")
df = pd.read_csv(url, dtype=dtype, parse_dates=parse_dates)

outside_jan = df[(df['tpep_pickup_datetime'] < '2021-01-01') | (df['tpep_pickup_datetime'] >= '2021-02-01')]

bad_fares = df[df['fare_amount'] <= 0]

bad_distances = df[df['trip_distance'] <= 0]

zero_passengers = df[df['passenger_count'] == 0]

duplicates_count = df.duplicated().sum()

problems_summary = pd.DataFrame({
    'Problem': [
        'Pickups outside Jan 2021',
        'Negative or Zero Fares (<= 0)',
        'Negative or Zero Distances (<= 0)',
        'Passenger Count = 0',
        'Missing Passenger Count (Nulls)',
        'Duplicate Rows'
    ],
    'Rows_Count': [
        len(outside_jan),
        len(bad_fares),
        len(bad_distances),
        len(zero_passengers),
        df['passenger_count'].isna().sum(),
        duplicates_count
    ]
})

print("\n--- Step 3: Problems Found ---")
print(problems_summary)
```

```python
def clean(df):
    cleaned_df = df.copy()
    
    rules_report = {}
    
    mask_null_passengers = cleaned_df['passenger_count'].isna()
    rules_report['Missing Passenger Count (Nulls)'] = mask_null_passengers.sum()
    cleaned_df = cleaned_df[~mask_null_passengers]
    
    mask_outside_jan = (cleaned_df['tpep_pickup_datetime'] < '2021-01-01') | (cleaned_df['tpep_pickup_datetime'] >= '2021-02-01')
    rules_report['Pickups outside Jan 2021'] = mask_outside_jan.sum()
    cleaned_df = cleaned_df[~mask_outside_jan]
    
    mask_bad_fares = cleaned_df['fare_amount'] <= 0
    rules_report['Negative or Zero Fares (<= 0)'] = mask_bad_fares.sum()
    cleaned_df = cleaned_df[~mask_bad_fares]
    
    return cleaned_df, rules_report

print(f"Rows before cleaning: {len(df)}")

cleaned_data, removal_stats = clean(df)

print(f"Rows after cleaning: {len(cleaned_data)}")

print("\nRows removed per rule:")
for rule, count in removal_stats.items():
    print(f"- {rule}: {count}")
```
